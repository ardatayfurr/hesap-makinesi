```python
def toplama():
    sayi1 = float(input("İlk sayıyı giriniz: "))
    sayi2 = float(input("2.Sayıyı Giriniz: "))
    sonuc = sayi1 + sayi2
    print(sonuc)

def çıkarma():
    sayi1 = float(input("İlk sayıyı giriniz: "))
    sayi2 = float(input("2.Sayıyı Giriniz: "))
    sonuc = sayi1 - sayi2
    print(sonuc)

def çarpma():
    sayi1 = float(input("İlk sayıyı giriniz: "))
    sayi2 = float(input("2.Sayıyı Giriniz: "))
    sonuc = sayi1 * sayi2
    print(sonuc)

def bölme():
    sayi1 = float(input("İlk sayıyı giriniz: "))
    sayi2 = float(input("2.Sayıyı Giriniz: "))
    sonuc = sayi1 / sayi2
    print(sonuc)

def hesap_makinesi():
    print("\n Hesap Makinesi Uygulaması")
    print("----------------------------")
    print("Lütfen yapmak istediğiniz işlemi seçiniz:")
    print("1: Toplama (+)")
    print("2: Çıkarma (-)")
    print("3: Çarpma (*)")
    print("4: Bölme (/)")
    print("5: Çıkış")
    print("----------------------------")
    secim = input("Seçiminizi giriniz (1/2/3/4/5): ")

    if secim == '1':
        toplama()
    elif secim == '2':
        çıkarma()
    elif secim == '3':
        çarpma()
    elif secim == '4':
        bölme()
    elif secim == '5':
        print("Hesap makinesi kapatılıyor. Güle güle!")
        return
    else:
        print("Geçersiz giriş. Lütfen sadece 1, 2, 3, 4 veya 5 giriniz.")
```
